```python
# Grava os arquivos do app (na pasta dados, só o relatos_inferencia.csv)
import base64, io, os, zipfile, importlib
with zipfile.ZipFile(io.BytesIO(base64.b64decode(APP_B64))) as z:
    for nome in z.namelist():
        destino = os.path.join(raiz, nome)
        os.makedirs(os.path.dirname(destino), exist_ok=True)
        with open(destino, "wb") as f:
            f.write(z.read(nome))
importlib.invalidate_caches()
PASTA_APP = os.path.join(raiz, "app_monitoramento")
for pasta, _, arquivos in sorted(os.walk(PASTA_APP)):
    print(pasta.replace(raiz, ""), sorted(arquivos))
assert os.path.exists(os.path.join(PASTA_APP, "motor", "relatos.py")), "o motor não foi gravado"
assert os.path.exists(os.path.join(PASTA_APP, "pages", "8_Plano_de_marketing.py")), "a página do marketing não foi gravada"
for arq in ["modelo_motor.json", "clientes_atendimento.csv", "ganho_teste.csv", "relatos_inferencia.csv"]:
    assert os.path.exists(os.path.join(PASTA_APP, "dados", arq)), f"falta dados/{arq}: rode o 10"
print("\nApp instalado em", PASTA_APP)
```
